# Fix SadTalker - Download Missing Safetensors

Safetensors are in `vinthony/SadTalker-V002rc`, NOT `vinthony/SadTalker`.

```python
import os
import json
import shutil
from huggingface_hub import hf_hub_download

MODEL_DIR = "/kaggle/working/sadtalker-model"
os.makedirs(MODEL_DIR, exist_ok=True)

downloads = [
    ("vinthony/SadTalker", "mapping_00109-model.pth.tar"),
    ("vinthony/SadTalker", "mapping_00229-model.pth.tar"),
    ("vinthony/SadTalker-V002rc", "SadTalker_V0.0.2_256.safetensors"),
    ("vinthony/SadTalker-V002rc", "SadTalker_V0.0.2_512.safetensors")
]

print("=== Downloading SadTalker ===")
for repo_id, file_name in downloads:
    target = os.path.join(MODEL_DIR, file_name)
    if os.path.exists(target):
        size_mb = os.path.getsize(target) / 1024**2
        print(f"{file_name} already exists ({size_mb:.1f}MB), skipping")
        continue
    print(f"Downloading {file_name} from {repo_id}...")
    try:
        hf_hub_download(repo_id=repo_id, filename=file_name, local_dir=MODEL_DIR)
        print("  OK")
    except Exception as e:
        print(f"  Error: {e}")
    cache_dir = os.path.expanduser('~/.cache/huggingface')
    if os.path.exists(cache_dir):
        shutil.rmtree(cache_dir)

# Create dataset-metadata.json
metadata = {
    "title": "sadtalker-model",
    "id": "kingtechie/sadtalker-model",
    "licenses": [{"name": "CC0-1.0"}]
}
with open(os.path.join(MODEL_DIR, "dataset-metadata.json"), "w") as f:
    json.dump(metadata, f, indent=4)
print("Created dataset-metadata.json")

# List files
print("\n=== All files in MODEL_DIR ===")
total = 0
for f in sorted(os.listdir(MODEL_DIR)):
    fp = os.path.join(MODEL_DIR, f)
    if os.path.isfile(fp):
        sz = os.path.getsize(fp)
        total += sz
        print(f"  {f} ({sz / 1024**2:.1f}MB)")
print(f"\nTotal: {total / 1024**3:.2f} GB")
```
