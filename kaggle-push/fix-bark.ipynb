# Fix Bark - Complete Re-download

Downloads ALL files from suno/bark needed for AutoProcessor + BarkModel.
Previous versions replaced the dataset. This uploads the complete set.

```python
import os
import json
import shutil
from huggingface_hub import hf_hub_download

MODEL_DIR = "/kaggle/working/bark-model"
os.makedirs(MODEL_DIR, exist_ok=True)

# All files needed for standard HuggingFace loading
files = [
    # Model files
    "config.json",
    "generation_config.json",
    "pytorch_model.bin",
    # Tokenizer files
    "tokenizer.json",
    "vocab.txt",
    "tokenizer_config.json",
    "special_tokens_map.json",
    "speaker_embeddings_path.json"
]

print("=== Downloading Bark TTS ===")
for file_name in files:
    target = os.path.join(MODEL_DIR, file_name)
    if os.path.exists(target):
        size_mb = os.path.getsize(target) / 1024**2
        print(f"{file_name} already exists ({size_mb:.1f}MB), skipping")
        continue
    print(f"Downloading {file_name}...")
    try:
        hf_hub_download(
            repo_id="suno/bark",
            filename=file_name,
            local_dir=MODEL_DIR
        )
        print("  OK")
    except Exception as e:
        print(f"  Error: {e}")
    cache_dir = os.path.expanduser('~/.cache/huggingface')
    if os.path.exists(cache_dir):
        shutil.rmtree(cache_dir)

# Create metadata
metadata = {
    "title": "bark-model",
    "id": "kingtechie/bark-model",
    "licenses": [{"name": "CC0-1.0"}]
}
with open(os.path.join(MODEL_DIR, "dataset-metadata.json"), "w") as f:
    json.dump(metadata, f, indent=4)

# List files
print("\n=== All files in MODEL_DIR ===")
total = 0
for f in sorted(os.listdir(MODEL_DIR)):
    fp = os.path.join(MODEL_DIR, f)
    if os.path.isfile(fp):
        sz = os.path.getsize(fp)
        total += sz
        print(f"  {f} ({sz / 1024**2:.1f}MB)")
print(f"\nTotal: {total / 1024**3:.2f} GB")
```
