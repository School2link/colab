# Download LatentSync to Kaggle Dataset

**Size:** ~8 GB | **Purpose:** Lip sync via audio-conditioned latent diffusion

### Instructions
1. Enable **Internet** (Settings gear > Internet > On)
2. Run all cells (Runtime > Run All)
3. Model auto-publishes as `kingtechie/latentsync-model`

```python
import os
import shutil
from huggingface_hub import hf_hub_download

MODEL_DIR = "/kaggle/working/latentsync-model"
os.makedirs(MODEL_DIR, exist_ok=True)

print("=== Downloading LatentSync (~8GB) ===")

all_files = [
    "latentsync_unet.pt",
    "latentsync_syncnet.pt",
    "config.json",
    "whisper/tiny.pt",
    "auxiliary/2DFAN4-cd938726ad.zip",
    "auxiliary/i3d_torchscript.pt",
    "auxiliary/koniq_pretrained.pkl",
    "auxiliary/s3fd-619a316812.pth",
    "auxiliary/sfd_face.pth",
    "auxiliary/syncnet_v2.model",
    "auxiliary/vgg16-397923af.pth",
    "auxiliary/vit_g_hybrid_pt_1200e_ssv2_ft.pth",
]

for file_name in all_files:
    print(f"Downloading {file_name}...")
    try:
        hf_hub_download(
            repo_id="ByteDance/LatentSync",
            filename=file_name,
            local_dir=MODEL_DIR
        )
        print("  OK")
    except Exception as e:
        print(f"  Error: {e}")
    cache_dir = os.path.expanduser('~/.cache/huggingface')
    if os.path.exists(cache_dir):
        shutil.rmtree(cache_dir)

total_size = 0
file_count = 0
for dirpath, dirnames, filenames in os.walk(MODEL_DIR):
    for f in filenames:
        fp = os.path.join(dirpath, f)
        total_size += os.path.getsize(fp)
        file_count += 1
print(f"\nDone! Files: {file_count}, Size: {total_size / 1024**3:.2f} GB")
```
