```python
import os

def get_dir_size(path):
    total = 0
    for dirpath, dirnames, filenames in os.walk(path):
        for f in filenames:
            fp = os.path.join(dirpath, f)
            if os.path.isfile(fp):
                total += os.path.getsize(fp)
    return total

def format_size(size_bytes):
    for unit in ['B', 'KB', 'MB', 'GB', 'TB']:
        if size_bytes < 1024:
            return f"{size_bytes:.2f} {unit}"
        size_bytes /= 1024
    return f"{size_bytes:.2f} PB"

print("=" * 60)
print("AVATAR MODELS")
print("=" * 60)
for root, dirs, files in os.walk("/kaggle/working/avatar-models"):
    level = root.replace("/kaggle/working/avatar-models", "").count(os.sep)
    indent = " " * 2 * level
    print(f"{indent}{os.path.basename(root)}/")
    subindent = " " * 2 * (level + 1)
    for file in sorted(files)[:10]:
        filepath = os.path.join(root, file)
        size = os.path.getsize(filepath)
        print(f"{subindent}{file}  ({format_size(size)})")
    if len(files) > 10:
        print(f"{subindent}... and {len(files) - 10} more files")

print("\n" + "=" * 60)
print("VISUAL MODELS")
print("=" * 60)
for root, dirs, files in os.walk("/kaggle/working/visual-models"):
    level = root.replace("/kaggle/working/visual-models", "").count(os.sep)
    indent = " " * 2 * level
    print(f"{indent}{os.path.basename(root)}/")
    subindent = " " * 2 * (level + 1)
    for file in sorted(files)[:10]:
        filepath = os.path.join(root, file)
        size = os.path.getsize(filepath)
        print(f"{subindent}{file}  ({format_size(size)})")
    if len(files) > 10:
        print(f"{subindent}... and {len(files) - 10} more files")

avatar_size = get_dir_size("/kaggle/working/avatar-models")
visual_size = get_dir_size("/kaggle/working/visual-models")
total_size = avatar_size + visual_size

print("\n" + "=" * 60)
print("SUMMARY")
print("=" * 60)
print(f"Avatar Models: {format_size(avatar_size)}")
print(f"Visual Models: {format_size(visual_size)}")
print(f"Total Size:    {format_size(total_size)}")
```
