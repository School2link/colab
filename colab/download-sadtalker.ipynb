# Download SadTalker to Kaggle Dataset

**Size:** ~2.5 GB | **Purpose:** Talking head animation

### Instructions
1. Enable **Internet** (Settings gear > Internet > On)
2. Run all cells (Runtime > Run All)
3. Model auto-publishes as `kingtechie/sadtalker-model`

```python
import os
import shutil
from huggingface_hub import hf_hub_download

MODEL_DIR = "/kaggle/working/sadtalker-model"
os.makedirs(MODEL_DIR, exist_ok=True)

print("=== Downloading SadTalker ===")
# Mapping files from vinthony/SadTalker, safetensors from vinthony/SadTalker-V002rc
downloads = [
    ("vinthony/SadTalker", "mapping_00109-model.pth.tar"),
    ("vinthony/SadTalker", "mapping_00229-model.pth.tar"),
    ("vinthony/SadTalker-V002rc", "SadTalker_V0.0.2_256.safetensors"),
    ("vinthony/SadTalker-V002rc", "SadTalker_V0.0.2_512.safetensors")
]
for repo_id, file_name in downloads:
    print(f"Downloading {file_name} from {repo_id}...")
    try:
        hf_hub_download(
            repo_id=repo_id,
            filename=file_name,
            local_dir=MODEL_DIR
        )
        print("  OK")
    except Exception as e:
        print(f"  Error: {e}")
    cache_dir = os.path.expanduser('~/.cache/huggingface')
    if os.path.exists(cache_dir):
        shutil.rmtree(cache_dir)

# Verify
total_size = 0
file_count = 0
for dirpath, dirnames, filenames in os.walk(MODEL_DIR):
    for f in filenames:
        fp = os.path.join(dirpath, f)
        total_size += os.path.getsize(fp)
        file_count += 1
print(f"Done! Files: {file_count}, Size: {total_size / 1024**3:.2f} GB")
```

```python
import kagglehub

DATASET_ID = "kingtechie/sadtalker-model"

print(f"\n=== Publishing as dataset: {DATASET_ID} ===")
kagglehub.dataset_upload(
    DATASET_ID,
    MODEL_DIR,
    version_notes="Initial download - SadTalker"
)
print(f"Published: https://www.kaggle.com/datasets/{DATASET_ID}")
```
