# Download SadTalker to Kaggle Dataset

**Size:** ~2.5 GB | **Purpose:** Talking head animation

### Instructions
1. Enable **Internet** (Settings gear > Internet > On)
2. Run all cells (Runtime > Run All)
3. Model auto-publishes as `kingtechie/sadtalker-model`

```python
import os
from huggingface_hub import snapshot_download

MODEL_DIR = "/kaggle/working/sadtalker-model"
os.makedirs(MODEL_DIR, exist_ok=True)

print("=== Downloading SadTalker (~2.5GB) ===")
snapshot_download(
    repo_id="camenduru/SadTalker",
    local_dir=MODEL_DIR,
    local_dir_use_symlinks=False
)

print("\n=== Verifying download ===")
total_size = 0
file_count = 0
for dirpath, dirnames, filenames in os.walk(MODEL_DIR):
    for f in filenames:
        fp = os.path.join(dirpath, f)
        total_size += os.path.getsize(fp)
        file_count += 1

print(f"Files: {file_count}")
print(f"Size: {total_size / 1024**3:.2f} GB")
```

```python
import kagglehub

DATASET_ID = "kingtechie/sadtalker-model"

print(f"\n=== Publishing as dataset: {DATASET_ID} ===")
kagglehub.dataset_upload(
    DATASET_ID,
    MODEL_DIR,
    version_notes="Initial download - SadTalker full"
)
print(f"Published: https://www.kaggle.com/datasets/{DATASET_ID}")
```
